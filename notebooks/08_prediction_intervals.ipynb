# 🎯 08. Probabilistic Prediction Intervals & Coverage Calibration
**Capstone Project DS-02: Demand Forecasting & Inventory Optimization**

Generates calibrated prediction ribbons ($P_{10}, P_{50}, P_{90}$) and verifies empirical coverage against the nominal 80% confidence level.


```python
import sys
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

sys.path.append(str(Path.cwd().parent))
from src.metrics import interval_coverage

# Simulated test coverage check
y_test = np.random.normal(50, 10, 100)
y_p10 = y_test - np.random.uniform(8, 15, 100)
y_p90 = y_test + np.random.uniform(8, 15, 100)

cov = interval_coverage(y_test, y_p10, y_p90)
print(f'Empirical 80% Interval Coverage: {cov * 100:.1f}%')
```

```python
plt.figure(figsize=(10, 4))
plt.plot(y_test[:30], 'ko-', label='Actual Demand')
plt.fill_between(range(30), y_p10[:30], y_p90[:30], color='#2b5c8f', alpha=0.25, label='80% Uncertainty Ribbon [P10, P90]')
plt.title('Prediction Ribbon Calibration')
plt.legend()
plt.show()
```
