# 🌳 07. Hierarchical Forecast Reconciliation (Darts & MinT)
**Capstone Project DS-02: Demand Forecasting & Inventory Optimization**

Reconciles incoherencies between store-level, state-level, and corporate-level forecasts using the **Minimum Trace (MinT-Shrink)** algorithm, guaranteeing 100% mathematical consistency across all echelons.


```python
import sys
from pathlib import Path
import numpy as np

sys.path.append(str(Path.cwd().parent))
from src.models.reconciliation import HierarchicalReconciler

rec = HierarchicalReconciler(method='mint_shrink')
print('Summing matrix S shape:', rec.S_matrix.shape)
print('Bottom nodes (stores):', rec.BOTTOM_NODES)
```

```python
# Simulate base incoherent predictions
h = 14
base_forecasts = {s: np.random.uniform(20, 35, h) for s in rec.BOTTOM_NODES}
base_forecasts['CA'] = sum(base_forecasts[s] for s in rec.BOTTOM_NODES if 'CA' in s) * 1.15
base_forecasts['TX'] = sum(base_forecasts[s] for s in rec.BOTTOM_NODES if 'TX' in s) * 0.90
base_forecasts['WI'] = sum(base_forecasts[s] for s in rec.BOTTOM_NODES if 'WI' in s) * 1.05
base_forecasts['Total'] = (base_forecasts['CA'] + base_forecasts['TX'] + base_forecasts['WI']) * 1.10

err_pre = rec.compute_coherency_error(base_forecasts)
print('Pre-reconciliation coherency error:', err_pre)

reconciled = rec.reconcile(base_forecasts)
err_post = rec.compute_coherency_error(reconciled)
print('Post-reconciliation coherency error:', err_post)
```
