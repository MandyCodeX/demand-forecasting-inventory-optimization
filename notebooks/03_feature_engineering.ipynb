# 🛠️ 03. Feature Engineering Pipeline: Autoregressive Lags & Rolling Statistics
**Capstone Project DS-02: Demand Forecasting & Inventory Optimization**

Builds a leak-free feature matrix for gradient boosting and deep learning models:
- Lags: 7, 14, 21, 28 days
- Rolling averages & standard deviations (shifted to preserve causality)
- Cyclical sine/cosine day of year & day of week encodings
- SNAP welfare assistance indicators
- Sell price volatility and markdown discount ratios


```python
import sys
from pathlib import Path
import pandas as pd
sys.path.append(str(Path.cwd().parent))
from src.data.data_loader import M5DataLoader
from src.features.build_features import FeaturePipeline

loader = M5DataLoader(data_dir='../data/sample')
merged_df = loader.build_merged_dataset(max_days=180)

pipeline = FeaturePipeline(lags=[7, 14, 21, 28], rolling_windows=[7, 14, 28])
featured_df = pipeline.fit_transform(merged_df)

print(f'Raw features: {merged_df.shape[1]} -> Engineered features: {featured_df.shape[1]}')
print('Sample columns:', [c for c in featured_df.columns if 'lag' in c or 'roll' in c])
```
