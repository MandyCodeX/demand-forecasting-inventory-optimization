**Capstone Project DS-02: Demand Forecasting & Inventory Optimization**

Solves the multi-period replenishment Mixed-Integer Linear Program (MILP) to minimize total inventory holding, stockout penalties, and batch dispatch fees.


```python
import sys
from pathlib import Path
import pandas as pd
import numpy as np

sys.path.append(str(Path.cwd().parent))
from src.optimizer import InventoryOptimizer, ItemParams

optimizer = InventoryOptimizer(planning_horizon_days=14, warehouse_capacity=10000, daily_procurement_budget=25000)
items = [
    ItemParams(item_id='FOODS_1_001', store_id='CA_1', unit_cost=3.50, lead_time_days=2, min_order_qty=15.0),
    ItemParams(item_id='HOUSEHOLD_1_001', store_id='CA_1', unit_cost=7.99, lead_time_days=3, min_order_qty=10.0),
    ItemParams(item_id='HOBBIES_1_001', store_id='CA_1', unit_cost=14.99, lead_time_days=2, min_order_qty=5.0),
]
```

```python
# Build forecast input
dates = pd.date_range('2016-05-01', periods=14, freq='D')
f_records = []
for item in items:
    for d in dates:
        f_records.append({'store_id': item.store_id, 'item_id': item.item_id, 'date': d, 'demand_p50': np.random.uniform(5, 18)})
f_df = pd.DataFrame(f_records)

res = optimizer.solve(f_df, items, demand_column='demand_p50')
print(f'Solver Status: {res.status}')
print(f'Total Cost: ${res.total_cost:,.2f}')
print(f'Service Level: {res.service_level:.1f}%')
print(f'Cost Reduction vs Heuristic: {res.comparison_vs_heuristic["percentage_saved"]:.1f}%')
res.orders_df.head(10)
```
