# 📈 02. Exploratory Data Analysis: Seasonality, Intermittency & SNAP Effects
**Capstone Project DS-02: Demand Forecasting & Inventory Optimization**

Explores retail time series patterns:
- Weekly day-of-week seasonality (weekend shopping spikes)
- Zero-inflated intermittency across slow-moving SKUs
- SNAP food stamp promotion lift across California, Texas, and Wisconsin
- Price markdown sensitivity


```python
import sys
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sys.path.append(str(Path.cwd().parent))
from src.data.data_loader import M5DataLoader

loader = M5DataLoader(data_dir='../data/sample')
merged_df = loader.build_merged_dataset(max_days=300)
print('Merged shape:', merged_df.shape)
merged_df.head()
```

```python
# 1. Aggregate demand trend over time
daily_sales = merged_df.groupby('date')['demand'].sum().reset_index()

plt.figure(figsize=(12, 4))
plt.plot(daily_sales['date'], daily_sales['demand'], color='#2b5c8f', lw=1.5)
plt.title('Nationwide Daily Total Sales (Sample Horizon)')
plt.xlabel('Date')
plt.ylabel('Total Units Sold')
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# 2. Day-of-week seasonality
merged_df['day_name'] = merged_df['date'].dt.day_name()
dow_order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']

plt.figure(figsize=(8, 4))
sns.barplot(data=merged_df, x='day_name', y='demand', order=dow_order, palette='Blues_d')
plt.title('Average Sales by Day of Week')
plt.ylabel('Average Units Sold')
plt.xticks(rotation=30)
plt.show()
```
